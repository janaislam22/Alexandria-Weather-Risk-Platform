```python
from sqlalchemy import create_engine
import pandas as pd
engine = create_engine(
    "mssql+pyodbc://localhost/Alex_weather"
    "?trusted_connection=yes"
    "&driver=ODBC+Driver+17+for+SQL+Server"
    "&TrustServerCertificate=yes"
)
```

```python
df = pd.read_sql( """
select * 
from bronze.weather """, con=engine)
```

```python
df.dtypes
```

```python
df['time']=pd.to_datetime(df['time'])
df['temperature_2m']=df['temperature_2m'].astype('float')
df['precipitation']=df['precipitation'].astype('float')
df['wind_gusts_10m']=df['wind_gusts_10m'].astype('float')
df['pressure_msl']=df['pressure_msl'].astype('float')
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates(subset=['location', 'time'])
```

```python
df.to_sql('weather', con=engine, schema='silver', if_exists='append', index=False)
```
