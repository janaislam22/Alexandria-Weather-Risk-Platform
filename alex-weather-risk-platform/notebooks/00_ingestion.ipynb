```python
import requests
import pandas as pd

loc = {"name": "Stanley", "lat": 31.24, "lon": 29.95}

url = "https://archive-api.open-meteo.com/v1/archive"
params = {
    "latitude": loc["lat"],
    "longitude": loc["lon"],
    "hourly": "temperature_2m,precipitation,wind_gusts_10m,pressure_msl",
    "timezone": "Africa/Cairo",
    "start_date": "2026-01-01",
    "end_date": "2026-03-31"
}
r = requests.get(url, params=params, timeout=30)
data = r.json()
df = pd.DataFrame(data["hourly"])
df["location"] = loc["name"]
df.to_csv("../data/weather_stanley.csv", index=False)
print(df.head(10))
```

```python
loc = {"name": "Borg_El_Arab", "lat": 30.85, "lon": 29.60}

url = "https://archive-api.open-meteo.com/v1/archive"
params = {
    "latitude": loc["lat"],
    "longitude": loc["lon"],
    "hourly": "temperature_2m,precipitation,wind_gusts_10m,pressure_msl",
    "timezone": "Africa/Cairo",
   "start_date": "2026-01-01",
   "end_date": "2026-03-31"
}
r = requests.get(url, params=params, timeout=30)
r.raise_for_status()
data = r.json()
df = pd.DataFrame(data["hourly"])
df["location"] = loc["name"]
df.to_csv("../data/weather_Borg_El_Arab.csv", index=False)
print(df.head(10))
```

```python
a = pd.read_csv("../data/weather_stanley.csv")
b = pd.read_csv("../data/weather_Borg_El_Arab.csv")
c = pd.read_csv("../data/weather_Abu_Qir.csv")
r.raise_for_status()
all_data = pd.concat([a, b, c], ignore_index=True)
all_data.to_csv("../data/weather_all.csv", index=False)

print(all_data.shape)
print(all_data["location"].value_counts())
```
